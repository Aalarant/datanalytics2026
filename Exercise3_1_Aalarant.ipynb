```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

In this exercise, we use the penguins dataset from seaborn.


```python
# With this line of code, we can load the penguins dataset from seaborn.
penguins = sns.load_dataset("penguins")
print(penguins.head())
```

# Let's create pair plot of the data.

```python
sns.pairplot(penguins.dropna())

```

# What correlations can we immediately see?

* From the pair plot, you'll typically notice, that bigger penguins have longer flippers.

# Let's use the island hue and see what kind of data it has.

```python
sns.pairplot(sns.load_dataset("penguins"), hue="island")
```

# What does this data reveal?

* Biscoe island dominates the dataset.
* Dream and Torgersen islands have fewer penguins.
* Some islands only have one type of penguin species.

# Now let's find the amount of penguins on each island by using pandas.

```python
penguins["island"].value_counts()
```

From this way of seeing the data we can see, that:
* Bisoce has the most amount of penguins.
* Torgersen has the least amount of penguins.

Therefore, we can say that Biscoe the island is "unbalanced".

# Let's create another pair plot, and use hue for the "species"- column.

```python
sns.pairplot(penguins, hue="species")
```

# Is there a difference in the distributions when using hue for the islands instead?
When comparing the two different hues with each other we can see that:
* Species separation is much clearer than island separation.
* Gentoo are larger.

# Now we create a scatter plot for bill length and flipper length, while using species as hue.

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.show()
```

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm", hue="island")
plt.show()
```

# Let's compare these different hue outcomes.
Looking at these two different data outputs we can see that:
* Species gives very clear separation, and Island doesn't.

# Now let's use box plots, violin plots or swarm plots to inspect the following information.
* flipper length
* bill length
* body mass
* Hue = island, x = species

```python
sns.violinplot(data=penguins, x="species", y="flipper_length_mm", hue="island", split=True)
plt.show()
```

```python
sns.violinplot(data=penguins, x="species", y="bill_length_mm", hue="island", split=True)
plt.show()
```

```python
sns.violinplot(data=penguins, x="species", y="body_mass_g", hue="island", split=True)
plt.show()
```

# From these three graphs we can see that
* Species differences dominate the dataset.
* Island influences species distribution.
