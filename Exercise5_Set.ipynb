# Exercise 3

```python
# Let's fetch the data from the API.
url = "https://edu.frostbit.fi/api/events/en"
response = requests.get(url)
data = response.json()

# Json phase
df_api = pd.json_normalize(
    data,
    record_path=["categories"],
    meta=[
        "id",
        "name",
        ["address", "street"],
        ["address", "city"],
        ["address", "zip"],
    ],
    errors="ignore",
    )

# Let's rename the columns to make them more readable.
df_api = df_api.rename(columns={"address.street": "street", "address.city": "city", "address.zip": "zip"})

# Show the results.
print(df_api.head())
df_api
```

# Exercise 4

```python
# Let's fetch the data from wikipedia.
url_rovaniemi = "https://en.wikipedia.org/wiki/Rovaniemi"
response = requests.get(url_rovaniemi)
soup = BeautifulSoup(response.content, "html.parser")


# Coordinates phase
coords_tag = soup.find("span",{"id": "coordinates"})

if coords_tag:
    coords = coords_tag.get_text(strip=True)
    print("Coordinates:", coords)
else: 
    print("Coordinates not found.")




# Nickname phase
nicknames = None
infobox = soup.find("table", {"class": "infobox"})

if infobox:
    for row in infobox.find_all("tr"):
        header = row.find("th")
        if header and "Nickname" in header.get_text():
            value_cell = row.find("td")
            if value_cell:
                nicknames = value_cell.get_text(strip=True)
                break

print("Nicknames:", nicknames)
           
```
