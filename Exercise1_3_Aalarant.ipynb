```python
# Task 2:
# Let's create a 7x7 array (matrix)
# With random integers between 0 and 49.
rng = numpy.random.default_rng()
data = rng.integers(0, 49, size = 49).reshape(7, 7)
print(data)
```

```python
# Task 3:
# Let's generate 8x8 random number matrix
# Integer values between 0 and 5.
data = rng.integers(0, 6, size=64).reshape(8, 8)
print(data)
```

```python
# Task 4:
# Let's generate 8x8 random number matrix
# Using standard normal distribution values.
# (Values are any decimal values between -3 and 3).
data = rng.integers(-3, 3, size=64).reshape(8,8)
print(data)
```
