```python
# Task 1:
# Let's create an list that displays fifteen one's
numbers = [1]

for x in range(0, 14):
    numbers.append(numbers[x])
print(numbers)
```

```python
# Task 2:
# Let's create an list that displays fifteen 7's using list comprehension.
numbers = [7 for x in range(0, 15)]
print(numbers)
```

```python
# Task 3:
# Let's create a list that display the numbers from 100 to 149 using list comprehension.
numbers = [x for x in range(100, 150)]
print(numbers)
```

```python
# Task 4:
# Let's create a list that displays even numbers between 0 and 100.
numbers = [x for x in range(0, 101) if x % 2 == 0]
print(numbers)
```

```python
# Task 5:
# Let's create a list that displays years between 1950 and 2020
# That are divisible by 4.
years = [x for x in range(1950, 2021) if x % 4 == 0]
print(years)
```
