<h3><b>Code: importing modules, custom functions and loading the data</b></h3>

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('motorcycle_sales.csv')
print(df.head())
```

<h3><b>Code: Data clean-up procedure</b></h3>

```python
# Show pairplot to visualize relationships between variables.
sns.pairplot(df, hue="year")
plt.show()
```

<h3><b>Code: Data analysis</b></h3>

```python
# code here. create more code cells / markdown cells if needed.
# see the examples on how code cells and markdown cells can be used
# at the same time for clear explanations

correlations = df.corr(numeric_only=True)
sns.heatmap(correlations, annot=True)
```

```python
# Let's see if there is any correlation between selling price and kilometers driven..
sns.boxplot(x='selling_price', y='km_driven', data=df)
```

<h3><b>Results and summary:</b></h3>

It seems that motorcycles that have more kilometers are cheaper than the newer models.
This dataset seems to be more focused on lower/mid priced motorcycles.

Also, selling price of the used vehicles closely follow the showroom price, which makes depreciation relatively predictable.
