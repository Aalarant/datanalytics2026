# Let's load the dataset.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("groceries.csv")
df.head()
```

# Let's hande these missing values by filling each with the mean of that column.

```python
# Pork Fish and Sunflower oil are missing values. Let's fill them with the mean of the column.
df.fillna({"Pork":df["Pork"].mean(), "Fish":df["Fish"].mean(), "Sunflower-oil":df["Sunflower-oil"].mean()}, inplace=True)
```

# Let's create a correlation matrix of the data. And also a heatmap on the correlations as well.

```python
numeric_df = df.select_dtypes(include=["number"])
corr = numeric_df.corr()
print(corr)
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation matrix of Groceries Dataset")
plt.tight_layout()
plt.show()
```

# Which grocery stands out?
Pork stands out the most.

# Which groceries correlate with each other's prices? What do they have in common?
Palm oil and corn, due to them having similar heatmap data.

# Let's use Histograms for Fish, Coffee and Tea.


```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
sns.histplot(df["Fish"], kde=True)
plt.title("Fish price distribution")

plt.subplot(1, 3, 2)
sns.histplot(df["Coffee"], kde=True)
plt.title("Coffee price distribution")

plt.subplot(1, 3, 3)
sns.histplot(df["Tea"], kde=True)
plt.title("Tea price distribution")

plt.tight_layout()
plt.show()
```

# Which prices have been historically typical for these products?
Let's see:
* Fish's price has been typically 5.
* Coffee's price has usually been around 2.
* And Tea's price has usually been 1.5.
