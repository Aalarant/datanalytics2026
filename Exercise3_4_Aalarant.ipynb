# Let's load and inspect the data.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("mobilephones.csv")
df.head()
```

# Regression plot.

```python
sns.lmplot(data=df, x="RAM", y="Price", height=6, aspect=1.5)
plt.title("RAM vs Price")
plt.show()
```

# What is the correlation based on the regression line?
Bigger Ram's usually cost a lot of money.

# Now let's do the regression plot: RAM VS PRICE with hue = brand.

```python
sns.lmplot(data=df, x="RAM", y="Price", hue="Brand", height=6, aspect=1.5)
plt.title("RAM vs Price by Brand")
plt.show()
```

# How is the correlation different when compared to without hue?
When you add hue:
* Each brand has their own regression line.
* Some brands cost alot more than other brands.

# Let's use regression plot for RAM and Price, no hue.

```python
sns.lmplot(data=df, x="BatteryCapacity", y="Price", height=6, aspect=1.5)
plt.title("Battery Capacity vs Price")
plt.show()
```

# What is the  correlation based on the regression line?
Battery capacity has much weaker correlation with price compared to RAM.

# Let's see the correlation with BatteryCapacity and Price with hue = brand.

```python
sns.lmplot(data=df, x="BatteryCapacity", y="Price", hue="Brand", height=6, aspect=1.5)
plt.title("Battery Capacity vs Price by Brand") 
plt.show()
```

# How is the correlation different when compared to without hue?
Brand differences become even more visible when we use hue.
