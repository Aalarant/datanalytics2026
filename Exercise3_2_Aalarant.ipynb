# Let's explore the MPG dataset from Seaborn.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

mpg = sns.load_dataset("mpg")
mpg.head()
```

# Let's clean up the data.


```python
# Let's create a new column called liters per 100km.
mpg["liters_per_100km"] = 235.215 / mpg["mpg"]
```

# Now let's remove the original mpg column and the name column.

```python
mpg = mpg.drop(columns=["mpg", "name"])
mpg.head()
```

# Let's create a correlation matrix.

```python
corr = mpg.corr(numeric_only=True)
corr
```

# Let's remove the columns model year and acceleration.
Why?
* They have negative values.

```python
mpg = mpg.drop(columns=["acceleration", "model_year"])
```

# Let's make a heatmap of the cleaned correlation matrix.

```python
sns.heatmap(mpg.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

# Let's pick one column related to the car's effiency.
Let's pick displacement, because:
* It correlates strongly with weight, horsepower, cylinders, and fuel consumption.
* It reflects the engine size.



```python
mpg = mpg.drop(columns=["cylinders", "horsepower"])
```

# Let's use pair plot with hue = origin.

```python
mpg = sns.load_dataset("mpg").dropna()
mpg["liters_per_100km"] = 235.215 / mpg["mpg"]

sns.pairplot(mpg, hue="origin")
plt.show()
```

# Which origin has higher or lower fuel consumption?
Generally:
* Usa has the highest fuel consumption, heaviest cars, and largest engines.
* Japan has the smallest engines, lightest cars, and best fuel economy.
* Europe has moderate engine sizes, and balanced effiency.

# What other features lead to higher or lower consumption?
Cars with higher consumption tend to have:
* Higher weight.
* Bigger engines.
* More horsepower.
* Lower acceleration.

And cars with lower consumption tend to have:
* Lower weight.
* Smaller engines.
* Less horsepower.
* Faster acceleration.
